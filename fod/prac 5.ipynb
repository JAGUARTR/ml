```python
import numpy as np
def best_rank_k_approx(A, k):
    U, S, VT = np.linalg.svd(A, full_matrices=False)
    S_k = np.zeros_like(S)
    S_k[:k] = S[:k]
    return U @ np.diag(S_k) @ VT

# Input matrix size
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))
print("Enter matrix row-wise:")
A = []
for i in range(rows):
    row = list(map(float, input().split()))
    A.append(row)
A = np.array(A)
k = int(input("Enter value of k: ")) # Fixed: Added closing double quote
A_k = best_rank_k_approx(A, k)
print("\nOriginal Matrix:\n", A)
print("\nBest Rank-k Approximation:\n", A_k)
```
